```python
import html
import pandas as pd
import ast
```

```python
extended_raw = pd.read_csv("../data/raw/extended_games.csv")
```

```python
extended_games = extended_raw.copy()
```

# extended_games.csv

```python
extended = extended_raw.drop_duplicates(subset=["item_id"])
extended = extended.dropna(subset=["item_id"])
```

```python
# Keep text-related columns
keep_cols_ext = [
    "item_id",
    "item_name",
    "short_description",
    "about_the_game",
    "detailed_description",
    "reviews",
    "tags",
    "genres",
    "categories",
    "supported_languages",
    "developers",
    "publishers",
]
extended = extended[keep_cols_ext].copy()
```

```python
def clean_list_str(x):
    """Clean text values."""
    if not isinstance(x, str) or x.strip() == "":
        return ""
    try:
        val = ast.literal_eval(x)
        if isinstance(val, list):
            return " ".join([str(v).lower().strip() for v in val])
    except:
        return x.lower().strip()
    return x.lower().strip()
```

```python
list_cols = ["tags", "genres", "categories", "supported_languages", "developers", "publishers"]

for col in list_cols:
    extended[col] = extended[col].apply(clean_list_str)
```

```python
# Basic cleaning for long text fields
text_cols = [
    "short_description",
    "about_the_game",
    "detailed_description",
    "reviews"
]

for col in text_cols:
    extended[col] = (
        extended[col].fillna("")
        .astype(str)
        .str.replace("\n", " ", regex=False)
        .str.replace("\r", " ", regex=False)
        .str.lower()
        .str.strip()
    )
```

```python
extended.to_csv("../data/cleansed/extended_games_cleansed_text.csv", index=False)

extended.head(5)
```
